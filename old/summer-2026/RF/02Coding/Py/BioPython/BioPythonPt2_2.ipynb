```python
# Lab: Biopython's Alignment Module
import sys

# Run this in Google Colab
!{sys.executable} -m pip install biopython

from Bio import Align
from Bio.Seq import Seq
from Bio.SeqRecord import SeqRecord
import pandas as pd

print("=" * 70)
print("LAB 2: BIOPYTHON'S ALIGNMENT MODULE (MODERN APPROACH)")
print("=" * 70)

# ===== PART 1: Introduction to Bio.Align =====
print("\n" + "=" * 70)
print("PART 1: Introduction to Bio.Align")
print("=" * 70)

print("\nBio.Align advantages:")
print("  - Modern, actively maintained")
print("  - Better performance")
print("  - Support for BLOSUM matrices")
print("  - Better visualization options")
print("  - Alignment objects with rich information")

# ===== PART 2: Basic Alignment with Bio.Align =====
print("\n" + "=" * 70)
print("PART 2: Basic Alignment with Bio.Align")
print("=" * 70)

# Create aligner object
aligner = Align.PairwiseAligner()

# Set alignment parameters
aligner.mode = 'global'  # global or local
aligner.match_score = 2
aligner.mismatch_score = -1
aligner.open_gap_score = -10
aligner.extend_gap_score = -0.5

# Sample sequences
seq1 = "ATGCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCG"
seq2 = "ATGCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATAA"

print(f"\nSequence 1: {seq1}")
print(f"Sequence 2: {seq2}")

# Perform alignment
alignments = aligner.align(seq1, seq2)

print(f"\nNumber of alignments: {len(alignments)}")
print(f"Best score: {alignments[0].score}")

# Display alignment
print(f"\nBest alignment:")
print(alignments[0])

# ===== PART 3: Alignment Object Properties =====
print("\n" + "=" * 70)
print("PART 3: Exploring Alignment Object Properties")
print("=" * 70)

alignment = alignments[0]

print(f"\nAlignment properties:")
print(f"  Score: {alignment.score}")
print(f"  Query sequence: {alignment.query}")
print(f"  Target sequence: {alignment.target}")
qb = alignment.aligned[0]
tb = alignment.aligned[1]
print(f"  Query start:  { qb[0][0]  }")
print(f"  Query end:    { qb[-1][-1]}")
print(f"  Target start: { tb[0][0]  }")
print(f"  Target end:   { tb[-1][-1]}")

# Get aligned sequences
print(f"\nAligned sequences:")
for i, (query_char, target_char) in enumerate(zip(alignment.query, alignment.target)):
    if i < 30:  # Print first 30 positions
        print(f"Position {i}: {query_char} - {target_char}")

# ===== PART 4: Local vs Global Alignment =====
print("\n" + "=" * 70)
print("PART 4: Local vs Global Alignment")
print("=" * 70)

# Create sequences with different regions
seq_a = "ATGCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCG"
seq_b = "AAAAAATGCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGAAAAAAA"

print(f"\nSequence A: {seq_a}")
print(f"Sequence B: {seq_b}")

# Global alignment
aligner_global = Align.PairwiseAligner()
aligner_global.mode = 'global'
aligner_global.match_score = 2
aligner_global.mismatch_score = -1
aligner_global.open_gap_score = -10
aligner_global.extend_gap_score = -0.5

global_aligns = aligner_global.align(seq_a, seq_b)
print(f"\n--- GLOBAL ALIGNMENT ---")
print(f"Score: {global_aligns[0].score}")
print(global_aligns[0])

# Local alignment
aligner_local = Align.PairwiseAligner()
aligner_local.mode = 'local'
aligner_local.match_score = 2
aligner_local.mismatch_score = -1
aligner_local.open_gap_score = -10
aligner_local.extend_gap_score = -0.5

local_aligns = aligner_local.align(seq_a, seq_b)
print(f"\n--- LOCAL ALIGNMENT ---")
print(f"Score: {local_aligns[0].score}")
print(local_aligns[0])

# ===== PART 5: Using BLOSUM Matrices for Proteins =====
print("\n" + "=" * 70)
print("PART 5: Using BLOSUM Matrices for Protein Alignment")
print("=" * 70)

# Import BLOSUM matrices
from Bio.Align import substitution_matrices

print("\nAvailable BLOSUM matrices:")
available_matrices = substitution_matrices.load()
print(f"  {', '.join(available_matrices[:5])}...")

# Load BLOSUM62 (most common)
blosum62 = substitution_matrices.load("BLOSUM62")

print(f"\nBLOSUM62 matrix shape: {blosum62.shape}")
print(f"Sample scores:")
print(f"  A-A: {blosum62['A', 'A']}")
print(f"  A-G: {blosum62['A', 'G']}")
print(f"  A-W: {blosum62['A', 'W']}")

# Create protein aligner with BLOSUM62
protein_aligner = Align.PairwiseAligner()
protein_aligner.substitution_matrix = blosum62
protein_aligner.open_gap_score = -10
protein_aligner.extend_gap_score = -0.5

# Sample proteins
prot1 = "MKVLWAALLVTFLAGCAKAKAVQVKVKALPDAQFEVVHSLAKWKRQTLGQHDFSAGEGLYTHMKALRPDEDRLSPLHSVYVDQWDWERVMGDGERQFSTLKSTVEAIWAGIKATEAAVSEEFGLAPFLPDQIHFVHSQELLSRYPDLDAKGRERAIAKDLGAVFLVGIGGKLSDGHRHDVRAPDYDDWSTPSELGHAGLNGDILVWNPVLEDAFELSSMGIRVDADTLKHQLALTGDEDRLELEWHQALLRGEMPQTIGGGIGQSRLTMLLLQLPHIGQVQAGVWPAAVRESVPSLL"
prot2 = "MKVLWAALLVTFLAGCAKAKAVQVKVKALPDAQFEVVHSLAKWKRQTLGQHDFSAGEGLYTHMKALRPDEDRLSPLHSVYVDQWDWERVMGDGERQFSTLKSTVEAIWAGIKATEAAVSEEFGLAPFLPDQIHFVHSQELLSRYPDLDAKGRERAIAKDLGAVFLVGIGGKLSDGHRHDVRAPDYDDWSTPSELGHAGLNGDILVWNPVLEDAFELSSMGIRVDADTLKHQLALTGDEDRLELEWHQALLRGEMPQTIGGGIGQSRLTMLLLQLPHIGQV"

print(f"\nProtein 1 length: {len(prot1)} aa")
print(f"Protein 2 length: {len(prot2)} aa")

# Align proteins
prot_aligns = protein_aligner.align(prot1, prot2)

print(f"\nNumber of alignments: {len(prot_aligns)}")
print(f"Best score: {prot_aligns[0].score}")

print(f"\nBest alignment (first 50 positions):")
alignment_str = str(prot_aligns[0]).split('\n')
for line in alignment_str[:3]:
    print(line[:80])

# ===== PART 6: Comparing Different Matrices =====
print("\n" + "=" * 70)
print("PART 6: Comparing Different Substitution Matrices")
print("=" * 70)

matrices_to_test = ["BLOSUM45", "BLOSUM62", "BLOSUM80", "PAM30", "PAM70"]
results = []

for matrix_name in matrices_to_test:
    try:
        matrix = substitution_matrices.load(matrix_name)
        aligner_test = Align.PairwiseAligner()
        aligner_test.substitution_matrix = matrix
        aligner_test.open_gap_score = -10
        aligner_test.extend_gap_score = -0.5

        aligns = aligner_test.align(prot1, prot2)
        score = aligns[0].score

        results.append({
            'Matrix': matrix_name,
            'Score': score,
            'Alignments': len(aligns)
        })
    except:
        print(f"  {matrix_name} not available")

print("\nMatrix comparison:")
df = pd.DataFrame(results)
print(df.to_string(index=False))

# ===== PART 7: Alignment Visualization =====
print("\n" + "=" * 70)
print("PART 7: Alignment Visualization and Formatting")
print("=" * 70)

def format_alignment(alignment, width=60):
    """Format alignment for better visualization"""
    query = str(alignment.query)
    target = str(alignment.target)

    print(f"\nAlignment Score: {alignment.score}")
    tb = alignment.aligned[0]
    qb = alignment.aligned[1]

    print(f"Query:  {qb[0][0]}-{qb[-1][-1]}")
    print(f"Target: {tb[0][0]}-{tb[-1][-1]}")
    print()

    # Print in chunks
    for i in range(0, len(query), width):
        print(f"Query:  {query[i:i+width]}")

        # Print match line
        match_line = ""
        for j in range(i, min(i+width, len(query))):
            if query[j] == target[j]:
                match_line += "|"
            else:
                match_line += " "
        print(f"        {match_line}")

        print(f"Target: {target[i:i+width]}")
        print()

# Format the best alignment
best_align = alignments[0]
format_alignment(best_align, width=50)

# ===== PART 8: Batch Alignment =====
print("\n" + "=" * 70)
print("PART 8: Batch Alignment of Multiple Sequences")
print("=" * 70)

sequences = [
    ("Seq1", "ATGCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCG"),
    ("Seq2", "ATGCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATAA"),
    ("Seq3", "ATGCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATGG"),
]

reference = sequences[0][1]
aligner_batch = Align.PairwiseAligner()
aligner_batch.mode = 'global'
aligner_batch.match_score = 2
aligner_batch.mismatch_score = -1
aligner_batch.open_gap_score = -10
aligner_batch.extend_gap_score = -0.5

print(f"\nReference sequence: {reference}")
print(f"\nAligning {len(sequences)-1} sequences to reference:")

batch_results = []
for name, seq in sequences[1:]:
    aligns = aligner_batch.align(reference, seq)
    score = aligns[0].score

    # Calculate identity
    query_aligned = str(aligns[0].query)
    target_aligned = str(aligns[0].target)
    matches = sum(1 for a, b in zip(query_aligned, target_aligned) if a == b)
    identity = (matches / len(query_aligned)) * 100

    batch_results.append({
        'Sequence': name,
        'Score': score,
        'Identity %': f"{identity:.2f}",
        'Length': len(seq)
    })

print("\nBatch alignment results:")
df_batch = pd.DataFrame(batch_results)
print(df_batch.to_string(index=False))

print("\n" + "=" * 70)
print("Lab 2 Complete!")
print("=" * 70)
```
