```python
# Lab 1: Pairwise Alignment with Bio.pairwise2

import sys

# Run this in Google Colab
!{sys.executable} -m pip install biopython

from Bio import pairwise2
from Bio.Seq import Seq
import pandas as pd

print("=" * 70)
print("LAB 1: PAIRWISE ALIGNMENT WITH BIO.PAIRWISE2")
print("=" * 70)

# ===== PART 1: Understanding Alignment Scoring =====
print("\n" + "=" * 70)
print("PART 1: Understanding Alignment Scoring Systems")
print("=" * 70)

# Define scoring matrices
print("\nCommon Scoring Matrices:")
print("  Match: +1 (identical bases/amino acids)")
print("  Mismatch: -1 (different bases/amino acids)")
print("  Gap opening: -10 (cost to open a gap)")
print("  Gap extension: -0.5 (cost per additional gap)")

# ===== PART 2: Simple DNA Sequence Alignment =====
print("\n" + "=" * 70)
print("PART 2: Simple DNA Sequence Alignment")
print("=" * 70)

# Create sample DNA sequences
seq1 = "ATGCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCG"
seq2 = "ATGCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATAA"

print(f"\nSequence 1: {seq1}")
print(f"Sequence 2: {seq2}")
print(f"Length seq1: {len(seq1)} bp")
print(f"Length seq2: {len(seq2)} bp")

# Perform global alignment (Needleman-Wunsch)
print("\n\n--- GLOBAL ALIGNMENT (Needleman-Wunsch) ---")
print("Aligns entire sequences end-to-end")

alignments_global = pairwise2.align.globalxx(seq1, seq2)

print(f"\nNumber of optimal alignments: {len(alignments_global)}")
print(f"Alignment score: {alignments_global[0][2]}")

# Display the best alignment
best_alignment = alignments_global[0]
print(f"\nBest alignment:")
print(f"Seq1: {best_alignment[0]}")
print(f"Seq2: {best_alignment[1]}")
print(f"Score: {best_alignment[2]}")

# Perform local alignment (Smith-Waterman)
print("\n\n--- LOCAL ALIGNMENT (Smith-Waterman) ---")
print("Aligns most similar regions")

alignments_local = pairwise2.align.localxx(seq1, seq2)

print(f"\nNumber of optimal alignments: {len(alignments_local)}")
print(f"Alignment score: {alignments_local[0][2]}")

best_local = alignments_local[0]
print(f"\nBest local alignment:")
print(f"Seq1: {best_local[0]}")
print(f"Seq2: {best_local[1]}")
print(f"Score: {best_local[2]}")

# ===== PART 3: Alignment with Custom Scoring =====
print("\n" + "=" * 70)
print("PART 3: Alignment with Custom Scoring Matrices")
print("=" * 70)

# Global alignment with match/mismatch scores
print("\n--- Global Alignment with Custom Scores ---")
print("Match: +2, Mismatch: -1")

alignments_custom = pairwise2.align.globalms(seq1, seq2, 2, -1, -10, -0.5)

print(f"Number of alignments: {len(alignments_custom)}")
print(f"Best score: {alignments_custom[0][2]}")

best_custom = alignments_custom[0]
print(f"\nAlignment:")
print(f"Seq1: {best_custom[0]}")
print(f"Seq2: {best_custom[1]}")
print(f"Score: {best_custom[2]}")

# Local alignment with custom scores
print("\n\n--- Local Alignment with Custom Scores ---")
print("Match: +3, Mismatch: -2, Gap open: -5, Gap extend: -1")

alignments_local_custom = pairwise2.align.localms(seq1, seq2, 3, -2, -5, -1)

print(f"Number of alignments: {len(alignments_local_custom)}")
print(f"Best score: {alignments_local_custom[0][2]}")

best_local_custom = alignments_local_custom[0]
print(f"\nAlignment:")
print(f"Seq1: {best_local_custom[0]}")
print(f"Seq2: {best_local_custom[1]}")
print(f"Score: {best_local_custom[2]}")

# ===== PART 4: Analyzing Alignment Quality =====
print("\n" + "=" * 70)
print("PART 4: Analyzing Alignment Quality")
print("=" * 70)

def analyze_alignment(seq1_aligned, seq2_aligned, score):
    """Analyze alignment statistics"""
    matches = sum(1 for a, b in zip(seq1_aligned, seq2_aligned) if a == b)
    mismatches = sum(1 for a, b in zip(seq1_aligned, seq2_aligned) if a != b and a != '-' and b != '-')
    gaps = sum(1 for a, b in zip(seq1_aligned, seq2_aligned) if a == '-' or b == '-')

    alignment_length = len(seq1_aligned)
    identity = (matches / alignment_length) * 100

    return {
        'matches': matches,
        'mismatches': mismatches,
        'gaps': gaps,
        'alignment_length': alignment_length,
        'identity': identity,
        'score': score
    }

# Analyze global alignment
stats_global = analyze_alignment(best_alignment[0], best_alignment[1], best_alignment[2])

print("\n--- Global Alignment Statistics ---")
print(f"Matches: {stats_global['matches']}")
print(f"Mismatches: {stats_global['mismatches']}")
print(f"Gaps: {stats_global['gaps']}")
print(f"Alignment length: {stats_global['alignment_length']}")
print(f"Identity: {stats_global['identity']:.2f}%")
print(f"Score: {stats_global['score']}")

# Analyze local alignment
stats_local = analyze_alignment(best_local[0], best_local[1], best_local[2])

print("\n--- Local Alignment Statistics ---")
print(f"Matches: {stats_local['matches']}")
print(f"Mismatches: {stats_local['mismatches']}")
print(f"Gaps: {stats_local['gaps']}")
print(f"Alignment length: {stats_local['alignment_length']}")
print(f"Identity: {stats_local['identity']:.2f}%")
print(f"Score: {stats_local['score']}")

# ===== PART 5: Protein Sequence Alignment =====
print("\n" + "=" * 70)
print("PART 5: Protein Sequence Alignment")
print("=" * 70)

# Sample protein sequences
prot1 = "MKVLWAALLVTFLAGCAKAKAVQVKVKALPDAQFEVVHSLAKWKRQTLGQHDFSAGEGLYTHMKALRPDEDRLSPLHSVYVDQWDWERVMGDGERQFSTLKSTVEAIWAGIKATEAAVSEEFGLAPFLPDQIHFVHSQELLSRYPDLDAKGRERAIAKDLGAVFLVGIGGKLSDGHRHDVRAPDYDDWSTPSELGHAGLNGDILVWNPVLEDAFELSSMGIRVDADTLKHQLALTGDEDRLELEWHQALLRGEMPQTIGGGIGQSRLTMLLLQLPHIGQVQAGVWPAAVRESVPSLL"
prot2 = "MKVLWAALLVTFLAGCAKAKAVQVKVKALPDAQFEVVHSLAKWKRQTLGQHDFSAGEGLYTHMKALRPDEDRLSPLHSVYVDQWDWERVMGDGERQFSTLKSTVEAIWAGIKATEAAVSEEFGLAPFLPDQIHFVHSQELLSRYPDLDAKGRERAIAKDLGAVFLVGIGGKLSDGHRHDVRAPDYDDWSTPSELGHAGLNGDILVWNPVLEDAFELSSMGIRVDADTLKHQLALTGDEDRLELEWHQALLRGEMPQTIGGGIGQSRLTMLLLQLPHIGQV"

print(f"\nProtein 1 length: {len(prot1)} aa")
print(f"Protein 2 length: {len(prot2)} aa")

# Align proteins
prot_alignments = pairwise2.align.globalms(prot1, prot2, 2, -1, -10, -0.5)

print(f"\nNumber of alignments: {len(prot_alignments)}")
print(f"Best score: {prot_alignments[0][2]}")

best_prot = prot_alignments[0]
print(f"\nBest alignment (first 50 characters):")
print(f"Prot1: {best_prot[0][:50]}")
print(f"Prot2: {best_prot[1][:50]}")

# Analyze protein alignment
prot_stats = analyze_alignment(best_prot[0], best_prot[1], best_prot[2])
print(f"\nProtein alignment identity: {prot_stats['identity']:.2f}%")
print(f"Score: {prot_stats['score']}")

# ===== PART 6: Comparing Different Scoring Schemes =====
print("\n" + "=" * 70)
print("PART 6: Comparing Different Scoring Schemes")
print("=" * 70)

seq_a = "ATGCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCG"
seq_b = "ATGCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATAA"

scoring_schemes = [
    ("Match +1, Mismatch -1", 1, -1, -10, -0.5),
    ("Match +2, Mismatch -1", 2, -1, -10, -0.5),
    ("Match +3, Mismatch -2", 3, -2, -10, -0.5),
    ("Match +5, Mismatch -3", 5, -3, -10, -0.5),
]

results = []

for name, match, mismatch, gap_open, gap_extend in scoring_schemes:
    aligns = pairwise2.align.globalms(seq_a, seq_b, match, mismatch, gap_open, gap_extend)
    score = aligns[0][2]
    stats = analyze_alignment(aligns[0][0], aligns[0][1], score)

    results.append({
        'Scoring Scheme': name,
        'Score': score,
        'Identity %': f"{stats['identity']:.2f}",
        'Matches': stats['matches'],
        'Mismatches': stats['mismatches']
    })

print("\nComparison of scoring schemes:")
df = pd.DataFrame(results)
print(df.to_string(index=False))

print("\n" + "=" * 70)
print("Lab 1 Complete!")
print("=" * 70)
```
