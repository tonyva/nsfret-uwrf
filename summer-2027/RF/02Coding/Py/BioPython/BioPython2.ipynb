```python
import sys
!{sys.executable} -m pip install biopython

from Bio import SeqIO
from Bio.SeqRecord import SeqRecord
from Bio.Seq import Seq
import os

# ===== PART 1: Creating Sample Files =====
print("=" * 60)
print("PART 1: Creating Sample Sequence Files")
print("=" * 60)

# Create a sample FASTA file
fasta_content = """>seq_001 | Escherichia coli gene A
ATGCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCG
ATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGAT
>seq_002 | Escherichia coli gene B
GATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGA
TCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCG
>seq_003 | Escherichia coli gene C
TCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATC
GATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGAT
"""

with open('sample_sequences.fasta', 'w') as f:
    f.write(fasta_content)
print("Created: sample_sequences.fasta")

# ===== PART 2: Reading FASTA Files =====
print("\n" + "=" * 60)
print("PART 2: Reading FASTA Files")
print("=" * 60)

# Method 1: Parse all sequences at once
print("\nMethod 1: Parse all sequences into a list")
sequences = list(SeqIO.parse('sample_sequences.fasta', 'fasta'))
print(f"Total sequences: {len(sequences)}")

for record in sequences:
    print(f"\nID: {record.id}")
    print(f"Description: {record.description}")
    print(f"Sequence length: {len(record.seq)} bp")
    print(f"First 30 bp: {record.seq[:30]}")

# Method 2: Iterate through file (memory efficient for large files)
print("\n\nMethod 2: Iterate through file (memory efficient)")
for record in SeqIO.parse('sample_sequences.fasta', 'fasta'):
    print(f"{record.id}: {len(record.seq)} bp")

# Method 3: Index sequences by ID (for random access)
print("\n\nMethod 3: Index sequences by ID")
seq_dict = SeqIO.index('sample_sequences.fasta', 'fasta')
print(f"Available sequences: {list(seq_dict.keys())}")
print(f"Accessing seq_002: {seq_dict['seq_002'].seq[:30]}")

# ===== PART 3: Reading GenBank Files =====
print("\n" + "=" * 60)
print("PART 3: Reading GenBank Files")
print("=" * 60)

# Create a sample GenBank file
genbank_content = """LOCUS       SAMPLE001                 120 bp    DNA     linear   UNA
DEFINITION  Sample DNA sequence for demonstration.
ACCESSION   SAMPLE001
VERSION     SAMPLE001.1
KEYWORDS    .
SOURCE      .
  ORGANISM  .
FEATURES             Location/Qualifiers
     source          1..120
                     /organism="Escherichia coli"
                     /mol_type="genomic DNA"
     gene            1..120
                     /gene="geneA"
     CDS             1..120
                     /gene="geneA"
                     /product="protein A"
ORIGIN
        1 atgcgatcga tcgatcgatc gatcgatcga tcgatcgatc gatcgatcga tcgatcgatc
       61 gatcgatcga tcgatcgatc gatcgatcga tcgatcgatc gatcgatcga tcgatcgatc
      121 g
//
"""

with open('sample_sequence.gb', 'w') as f:
    f.write(genbank_content)
print("Created: sample_sequence.gb")

# Read GenBank file
print("\nReading GenBank file:")
gb_record = SeqIO.read('sample_sequence.gb', 'genbank')
print(f"Locus: {gb_record.name}")
print(f"Description: {gb_record.description}")
print(f"Sequence length: {len(gb_record.seq)} bp")
print(f"Annotations: {gb_record.annotations}")

# ===== PART 4: Writing Sequences to Files =====
print("\n" + "=" * 60)
print("PART 4: Writing Sequences to Files")
print("=" * 60)

# Create new SeqRecords
new_records = [
    SeqRecord(Seq("ATGCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCG"),
              id="new_gene_1", description="New gene 1", annotations={"molecule_type": "DNA"}),
    SeqRecord(Seq("GATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGA"),
              id="new_gene_2", description="New gene 2", annotations={"molecule_type": "DNA"}),
    SeqRecord(Seq("TCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATC"),
              id="new_gene_3", description="New gene 3", annotations={"molecule_type": "DNA"})
]

# Write to FASTA
output_fasta = "output_sequences.fasta"
SeqIO.write(new_records, output_fasta, "fasta")
print(f"Wrote {len(new_records)} sequences to {output_fasta}")

# Write to GenBank
output_gb = "output_sequences.gb"
SeqIO.write(new_records, output_gb, "genbank")
print(f"Wrote {len(new_records)} sequences to {output_gb}")

# ===== PART 5: Converting Between Formats =====
print("\n" + "=" * 60)
print("PART 5: Converting Between Formats")
print("=" * 60)

# Read FASTA and convert to GenBank
print("\nConverting FASTA to GenBank format:")
fasta_records_for_gb = []
for record in SeqIO.parse('sample_sequences.fasta', 'fasta'):
    record.annotations['molecule_type'] = 'DNA'  # Add the missing annotation
    fasta_records_for_gb.append(record)

converted_count = SeqIO.write(fasta_records_for_gb, 'converted_sequences.gb', 'genbank')
print(f"Converted {converted_count} sequences from FASTA to GenBank")

# ===== PART 6: Filtering and Manipulating Records =====
print("\n" + "=" * 60)
print("PART 6: Filtering and Manipulating Records")
print("=" * 60)

# Filter sequences by length
print("\nFiltering sequences by length (>100 bp):")
long_sequences = [rec for rec in SeqIO.parse('sample_sequences.fasta', 'fasta')
                  if len(rec.seq) > 100]
print(f"Found {len(long_sequences)} sequences longer than 100 bp")

# Modify records and write
print("\nModifying records (adding annotations):")
modified_records = []
for record in SeqIO.parse('sample_sequences.fasta', 'fasta'):
    record.annotations['date'] = '2024-01-15'
    record.annotations['organism'] = 'Escherichia coli'
    modified_records.append(record)

SeqIO.write(modified_records, 'annotated_sequences.fasta', 'fasta')
print(f"Wrote {len(modified_records)} annotated sequences to annotated_sequences.fasta")

# ===== PART 7: Handling Multiple Formats =====
print("\n" + "=" * 60)
print("PART 7: Handling Multiple Formats")
print("=" * 60)

# List supported formats
print("\nCommon SeqIO formats:")
print("  - fasta: FASTA format")
print("  - genbank: GenBank format")
print("  - fastq: FASTQ format (with quality scores)")
print("  - phylip: PHYLIP format")
print("  - stockholm: Stockholm alignment format")
print("  - clustal: Clustal alignment format")

print("\n" + "=" * 60)
print("Lab 2 Complete!")
print("=" * 60)
```
