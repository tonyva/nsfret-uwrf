```python
import sys
!{sys.executable} -m pip install biopython

from Bio.Seq import Seq
from Bio.SeqRecord import SeqRecord

# ===== PART 1: Creating and Using Seq Objects =====
print("=" * 60)
print("PART 1: Seq Objects")
print("=" * 60)

# Create a Seq object from a DNA sequence string
dna_seq = Seq("ATGCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCGATCG")
print(f"DNA Sequence: {dna_seq}")
print(f"Type: {type(dna_seq)}")
print(f"Length: {len(dna_seq)} bp")

# Seq objects support indexing and slicing
print(f"\nFirst 10 bases: {dna_seq[:10]}")
print(f"Last 10 bases: {dna_seq[-10:]}")
print(f"Every 3rd base: {dna_seq[::3]}")

# String concatenation
seq1 = Seq("ATGC")
seq2 = Seq("GATC")
combined = seq1 + seq2
print(f"\nConcatenation: {seq1} + {seq2} = {combined}")

# Repetition
repeated = seq1 * 3
print(f"Repetition: {seq1} * 3 = {repeated}")

# ===== PART 2: Sequence Methods =====
print("\n" + "=" * 60)
print("PART 2: Seq Object Methods")
print("=" * 60)

# Complement and reverse complement
print(f"Original: {dna_seq[:20]}")
print(f"Complement: {dna_seq[:20].complement()}")
print(f"Reverse complement: {dna_seq[:20].reverse_complement()}")

# Transcription (DNA to RNA)
rna_seq = dna_seq.transcribe()
print(f"\nDNA: {dna_seq[:20]}")
print(f"RNA: {rna_seq[:20]}")

# Back-transcription (RNA to DNA)
back_to_dna = rna_seq.back_transcribe()
print(f"Back-transcribed: {back_to_dna[:20]}")

# ===== PART 3: SeqRecord Objects =====
print("\n" + "=" * 60)
print("PART 3: SeqRecord Objects")
print("=" * 60)

# Create a SeqRecord with metadata
record = SeqRecord(
    seq=dna_seq,
    id="seq_001",
    name="sample_sequence",
    description="A sample DNA sequence for demonstration",
    annotations={
        "organism": "Escherichia coli",
        "date": "2024-01-15",
        "source": "Lab experiment"
    }
)

print(f"ID: {record.id}")
print(f"Name: {record.name}")
print(f"Description: {record.description}")
print(f"Sequence length: {len(record.seq)} bp")
print(f"Annotations: {record.annotations}")

# Access sequence from record
print(f"\nSequence from record: {record.seq[:20]}")

# Modify annotations
record.annotations["researcher"] = "Dr. Smith"
print(f"Updated annotations: {record.annotations}")

# ===== PART 4: Working with Multiple Records =====
print("\n" + "=" * 60)
print("PART 4: Multiple SeqRecords")
print("=" * 60)

# Create multiple records
records = [
    SeqRecord(Seq("ATGCGATCG"), id="gene1", description="Gene 1"),
    SeqRecord(Seq("GATCGATCG"), id="gene2", description="Gene 2"),
    SeqRecord(Seq("TCGATCGAT"), id="gene3", description="Gene 3")
]

print(f"Number of records: {len(records)}")
for rec in records:
    print(f"  {rec.id}: {len(rec.seq)} bp - {rec.description}")
```
